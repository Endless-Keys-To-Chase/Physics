```python
A1 = np.array([31.5,26.5,21.5,16.5,11.5])
A2 = np.array([[0.0074,0.0069,0.0072],[0.0061,0.0059,0.0059],[0.0048,0.0044,0.0046],[0.0034,0.0025,0.0031],[0.0020,0.0014,0.0017]])
radius = 0.311/(2000)
length = 25.4/1000

mean_k = [np.mean(A2[0,:]),np.mean(A2[1,:]),np.mean(A2[2,:]),np.mean(A2[3,:]),np.mean(A2[4,:])]
#v/t = (pi*P*r^4)/8*eta*length
values = []
print(mean_k)
i = 0
while i<5:
    eta = np.prod([np.pi,A1[i],np.pow(10,4),np.pow(radius,4)])/np.prod([8,length,mean_k[i],5,0.000001])
    print(f"{eta:.4f}")
    values.append(eta)
    i += 1

print(values)
print("Final coefficient of Viscosity according to calculations are : " ,f"{np.mean(values):.4f}")
```

```python
#to determine the delta(1/t) error for precise calculation of values with +/- errors.
i = 0
std_err = []
while i<5:
    x = np.std(A2[i,:])
    print("Standard deviation for : ",i+1,x)
    print("Standard error for ",i+1,f"{x/np.sqrt(3):.4f}")
    std_err.append(x/np.sqrt(3))
    i += 1
```

```python
#complex error
delta_eta_list = []
for o in range(5):
    # Direct propagation using relative error: delta_eta = eta * (std_err / mean_k)
    d_eta = values[o] * (std_err[o] / mean_k[o])
    delta_eta_list.append(d_eta)
    print(f"Set {o+1}: eta = {values[o]:.4f} ± {d_eta:.4f}")
```

```python
#Absolute error code
# Absolute deviation for each element from its row mean
abs_deviations = np.abs(A2 - np.mean(A2, axis=1, keepdims=True))

# Mean Absolute Error for each of the 5 sets
mae_per_set = np.mean(abs_deviations, axis=1)

#print("Mean Absolute Error for each set:", mae_per_set)
for i in range(0,5):
    print("Mean Absolute Error for each set:", f"{mae_per_set[i]:.4f}") 
```

```python
# NEW CODE
import numpy as np
Ball1_a = np.array([1.53,1.09,1.05,1.24,1.10,1.16,1.07,1.06])
Ball1_b = np.array([1.53,1.18,0.86,1.26,0.97,1.17,1,1.16])
Ball1_c = np.array([1.59,1.10,1.20,1.15,1.04,1.20,0.97,1.20])
Ball1_d = np.array([1.10,0.99,1.08,1.08,1.16,1.10,1.15,1.11])
g = 9.8
rho = 960
Zero_Err = -0.0001

Diameter_1 = 6.23*0.0001 -(Zero_Err)
Mass_1 = 1.0295/1000
#density  =  mass/(4/3pir^3)
sigma_1 = (Mass_1*(6))/(np.pi*Diameter_1**3)
Vt_1 = np.array([0.09237,0.08771,0.09070,0.08850])
# To determine eta we need vt = 2/9(a^2g/eta sigma - rho)
eta_1 = ((2/9)*(((Diameter_1*0.5)**2)*g)/Vt_1)*(sigma_1 - rho)

'''Data - 2 written down'''
Ball2_a = np.array([2.92,1.64,1.73,1.75,1.76,1.72,1.84,1.60])
Ball2_b = np.array([2.57,1.77,1.88,1.54,1.76,1.64,1.81,1.66])
Ball2_c = np.array([1.43,1.77,1.61,1.71,1.68,1.88,1.62,1.62])
Ball2_d = np.array([1.22,1.76,1.69,1.89,1.59,1.66,1.75,1.54])

Diameter_2 = 4.66*0.0001 -(Zero_Err)
Mass_2 = 0.4711/1000
#density  =  mass/(4/3pir^3)
sigma_2 = (Mass_2*(6))/(np.pi*Diameter_2**3)
Vt_2 = np.array([0.05780,0.05822,0.05822,0.06116])
# To determine eta we need vt = 2/9(a^2g/eta sigma - rho)
eta_2 = ((2/9)*(((Diameter_2*0.5)**2)*g)/Vt_2)*(sigma_2 - rho)

'''Data - 3 written down'''
Ball3_a = np.array([1.86,2.26,2.40,2.56,2.02,2.35,2.09,2.17])
Ball3_b = np.array([1.80,2.30,2.31,2.30,2.36,2.22,2.32,2.22])
Ball3_c = np.array([2.36,2.87,3.00,2.83,2.52,2.73,2.45,2.12])
Ball3_d = np.array([1.91,2.39,2.23,2.29,2.37,2.28,2.19,2.22])

Diameter_3 = 3.88*0.0001 -(Zero_Err)
Mass_3 = 0.2588/1000
#density  =  mass/(4/3pir^3)
sigma_3 = (Mass_3*(6))/(np.pi*Diameter_3**3)
Vt_3 = np.array([0.04717,0.04386,0.04073,0.04415])
# To determine eta we need vt = 2/9(a^2g/eta sigma - rho)
eta_3 = ((2/9)*(((Diameter_3*0.5)**2)*g)/Vt_3)*(sigma_3 - rho)

print(eta_1,eta_2,eta_3)
print("Mean individual Viscosity ",np.mean(eta_1),np.mean(eta_2),np.mean(eta_3))
print("Final Viscosity ",np.mean([np.mean(eta_1),np.mean(eta_2),np.mean(eta_3)]))
```

```python
#Error Propagation
#for this method error would be propagated in way that std_err(eta) = eta*(std_err(terminal velocity)/terminal Velocity)
vt__1a = 0.1/Ball1_a[4:]
vt__1b = 0.1/Ball1_b[4:]
vt__1c = 0.1/Ball1_c[4:]
vt__1d = 0.1/Ball1_d[4:]
vt__2a = 0.1/Ball2_a[4:]
vt__2b = 0.1/Ball2_b[4:]
vt__2c = 0.1/Ball2_c[4:]
vt__2d = 0.1/Ball2_d[4:]
vt__3a = 0.1/Ball3_a[4:]
vt__3b = 0.1/Ball3_b[4:]
vt__3c = 0.1/Ball3_c[4:]
vt__3d = 0.1/Ball3_d[4:]

vt_mean = np.array([np.mean(vt__1a),np.mean(vt__1b),np.mean(vt__1c),np.mean(vt__1d),np.mean(vt__2a),
                   np.mean(vt__2b),np.mean(vt__2c),np.mean(vt__2d),np.mean(vt__3a),np.mean(vt__3b),
                   np.mean(vt__3c),np.mean(vt__3d)])

std_err = np.array([np.std(vt__1a,ddof = 1)/np.sqrt(4),np.std(vt__1b,ddof = 1)/np.sqrt(4),np.std(vt__1c,ddof = 1)/np.sqrt(4),np.std(vt__1d,ddof = 1)/np.sqrt(4),np.std(vt__2a,ddof = 1)/np.sqrt(4),
                   np.std(vt__2b,ddof = 1)/np.sqrt(4),np.std(vt__2c,ddof = 1)/np.sqrt(4),np.std(vt__2d,ddof = 1)/np.sqrt(4),np.std(vt__3a,ddof = 1)/np.sqrt(4),np.std(vt__3b,ddof = 1)/np.sqrt(4),
                   np.std(vt__3c,ddof = 1)/np.sqrt(4),np.std(vt__3d,ddof = 1)/np.sqrt(4)])
print(std_err)
eta = np.concatenate([eta_1,eta_2,eta_3])
eta_err = eta*(std_err/vt_mean)
print(eta_err)
```

```python
#Absolute Error
def MER(A):
    return np.mean(np.abs(A - np.mean(A)))

Absolute_err = np.array([MER(vt__1a),MER(vt__1b),MER(vt__1c),MER(vt__1d),MER(vt__2a),MER(vt__2b),
                        MER(vt__2c),MER(vt__2d),MER(vt__3a),MER(vt__3b),MER(vt__3c),MER(vt__3d)])

Mean_absolute_err = np.array([np.mean([Absolute_err[0:4]]),np.mean([Absolute_err[4:8]]),np.mean([Absolute_err[8:12]])])
print("Mean Absolute Error in Terminal Terminal Velocity ",Mean_absolute_err)
print("Absolute Error in Terminal Velocity ",Absolute_err)
```
